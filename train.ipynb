```python
import shutil
import random
from pathlib import Path
from ultralytics import YOLO

def train_yolo_yolodataset(
    dataset_path="YOLODataset",
    model_size="n",  # n, s, m, l, x
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    project="runs",
    name="smoke_detection",
    train_split=0.8,  # 80% train, 20% val
    val_split=0.2,    # reszta to test (jeśli train+val < 1.0)
    patience=50,
    save=True,
    plots=True,
    seed=42
):
    """
    Trenuje model YOLO na zbiorze danych YOLODataset z automatycznym podziałem na train/val/test.
    
    Parametry:
    ----------
    dataset_path : str
        Ścieżka do folderu YOLODataset (domyślnie: "YOLODataset")
    model_size : str
        Rozmiar modelu: 'n' (nano), 's' (small), 'm' (medium), 'l' (large), 'x' (xlarge)
        (domyślnie: "n")
    epochs : int
        Liczba epok treningu (domyślnie: 100)
    imgsz : int
        Rozmiar obrazu wejściowego (domyślnie: 640)
    batch : int
        Rozmiar batcha (domyślnie: 16)
    device : int
        ID urządzenia (0 dla GPU, 'cpu' dla CPU, domyślnie: 0)
    project : str
        Ścieżka do projektu (domyślnie: "runs")
    name : str
        Nazwa eksperymentu (domyślnie: "smoke_detection")
    train_split : float
        Proporcja danych treningowych (domyślnie: 0.8)
    val_split : float
        Proporcja danych walidacyjnych (domyślnie: 0.2)
    patience : int
        Liczba epok bez poprawy przed wczesnym zatrzymaniem (domyślnie: 50)
    save : bool
        Czy zapisywać checkpointy (domyślnie: True)
    plots : bool
        Czy generować wykresy (domyślnie: True)
    seed : int
        Ziarno losowości dla reprodukowalności (domyślnie: 42)
    
    Zwraca:
    -------
    model : YOLO
        Wytrenowany model YOLO
    results : dict
        Wyniki treningu
    """
    
    # Ustaw ziarno losowości
    random.seed(seed)
    
    # Sprawdź czy ścieżka do zbioru istnieje
    dataset_path = Path(dataset_path)
    if not dataset_path.exists():
        raise FileNotFoundError(f"Folder ze zbiorem danych nie istnieje: {dataset_path}")
    
    images_dir = dataset_path / "images"
    labels_dir = dataset_path / "labels"
    
    if not images_dir.exists():
        raise FileNotFoundError(f"Folder images nie istnieje: {images_dir}")
    if not labels_dir.exists():
        raise FileNotFoundError(f"Folder labels nie istnieje: {labels_dir}")
    
    # Pobierz listę wszystkich obrazów
    image_files = list(images_dir.glob("*.png")) + list(images_dir.glob("*.jpg"))
    image_files = sorted(image_files)
    
    if len(image_files) == 0:
        raise ValueError(f"Nie znaleziono żadnych obrazów w: {images_dir}")
    
    print(f"📊 Znaleziono {len(image_files)} obrazów")
    
    # Sprawdź czy dane są już podzielone
    train_dir = dataset_path / "train" / "images"
    val_dir = dataset_path / "valid" / "images"
    test_dir = dataset_path / "test" / "images"
    
    has_test_split = False
    
    if train_dir.exists() and val_dir.exists():
        print("✅ Dane są już podzielone na train/val, używam istniejących folderów")
        has_test_split = test_dir.exists()
    else:
        print("📂 Dzielę dane na train/val/test...")
        
        # Losowo przetasuj dane
        random.shuffle(image_files)
        
        # Oblicz indeksy podziału
        n_total = len(image_files)
        n_train = int(n_total * train_split)
        n_val = int(n_total * val_split)
        
        train_files = image_files[:n_train]
        val_files = image_files[n_train:n_train + n_val]
        test_files = image_files[n_train + n_val:] if (train_split + val_split) < 1.0 else []
        has_test_split = len(test_files) > 0
        
        print(f"   - Train: {len(train_files)} ({len(train_files)/n_total*100:.1f}%)")
        print(f"   - Val: {len(val_files)} ({len(val_files)/n_total*100:.1f}%)")
        if test_files:
            print(f"   - Test: {len(test_files)} ({len(test_files)/n_total*100:.1f}%)")
        
        # Utwórz foldery
        for split_name, files in [("train", train_files), ("valid", val_files), ("test", test_files)]:
            if not files:
                continue
                
            split_images_dir = dataset_path / split_name / "images"
            split_labels_dir = dataset_path / split_name / "labels"
            
            split_images_dir.mkdir(parents=True, exist_ok=True)
            split_labels_dir.mkdir(parents=True, exist_ok=True)
            
            # Skopiuj pliki
            for img_file in files:
                # Skopiuj obraz
                shutil.copy2(img_file, split_images_dir / img_file.name)
                
                # Skopiuj odpowiadający plik label
                label_file = labels_dir / (img_file.stem + ".txt")
                if label_file.exists():
                    shutil.copy2(label_file, split_labels_dir / label_file.name)
                else:
                    print(f"⚠️  Ostrzeżenie: Brak pliku label dla {img_file.name}")
        
        print("✅ Podział danych zakończony")
    
    # Wczytaj klasy z pliku classes.txt
    classes_file = dataset_path / "classes.txt"
    if classes_file.exists():
        with open(classes_file, 'r', encoding='utf-8') as f:
            class_names = [line.strip() for line in f.readlines() if line.strip()]
    else:
        # Spróbuj odczytać z dataset.yaml
        yaml_file = dataset_path / "dataset.yaml"
        if yaml_file.exists():
            import yaml
            with open(yaml_file, 'r', encoding='utf-8') as f:
                yaml_data = yaml.safe_load(f)
                class_names = yaml_data.get('names', ['smoke'])
        else:
            class_names = ['smoke']  # Domyślna klasa
    
    num_classes = len(class_names)
    print(f"📋 Liczba klas: {num_classes} - {class_names}")
    
    # Utwórz plik data.yaml
    train_images_path = (dataset_path / "train" / "images").absolute().as_posix()
    val_images_path = (dataset_path / "valid" / "images").absolute().as_posix()
    
    yaml_content = f"""# Dataset configuration for YOLO training
path: {dataset_path.absolute().as_posix()}
train: {train_images_path}
val: {val_images_path}
"""
    
    if has_test_split:
        test_images_path = (dataset_path / "test" / "images").absolute().as_posix()
        yaml_content += f"test: {test_images_path}\n"
    
    yaml_content += f"""
nc: {num_classes}
names: {class_names}
"""
    
    # Zapisz plik YAML
    data_yaml = dataset_path / "data.yaml"
    with open(data_yaml, 'w', encoding='utf-8') as f:
        f.write(yaml_content)
    
    print(f"✅ Plik data.yaml utworzony: {data_yaml}")
    
    # Wybierz model
    model_name = "yolo11s.pt"
    print(f"📦 Ładowanie modelu {model_name}...")
    model = YOLO(model_name)
    
    print(f"🚀 Rozpoczynam trening...")
    print(f"   - Zbiór danych: {dataset_path}")
    print(f"   - Model: {model_name}")
    print(f"   - Epoki: {epochs}")
    print(f"   - Rozmiar obrazu: {imgsz}")
    print(f"   - Batch: {batch}")
    print(f"   - Urządzenie: {device}")
    print(f"   - Klasy: {class_names}")
    
    # Trenuj model
    yaml_path = data_yaml.absolute().as_posix()
    
    try:
        results = model.train(
            data=yaml_path,
            epochs=epochs,
            imgsz=imgsz,
            batch=batch,
            device=device,
            project=project,
            name=name,
            patience=patience,
            save=save,
            plots=plots,
            verbose=True
        )
        
        print(f"✅ Trening zakończony!")
        print(f"📊 Wyniki zapisane w: {project}/{name}")
        
        return model, results
    
    except Exception as e:
        print(f"❌ Błąd podczas treningu: {str(e)}")
        print(f"   Typ błędu: {type(e).__name__}")
        raise

print("✅ Funkcja train_yolo_yolodataset została załadowana!")
```

```python
# Przykład użycia funkcji dla YOLODataset
# Funkcja automatycznie podzieli dane na train/val/test i wytrenuje model

model, results = train_yolo_yolodataset(
    dataset_path="YOLODataset",
    model_size="s",      # 'n' (nano), 's' (small), 'm' (medium), 'l' (large), 'x' (xlarge)
    epochs=30,           # Liczba epok
    imgsz=640,            # Rozmiar obrazu (512, 640, 1280, itp.)
    batch=20,             # Rozmiar batcha (dostosuj do pamięci GPU)
    device=0,             # 0 dla GPU, 'cpu' dla CPU
    project="runs",
    name="smoke_detection",
    train_split=0.8,      # 80% train
    val_split=0.2,        # 20% val (reszta to test jeśli train+val < 1.0)
    patience=10,           # Early stopping
    save=True,
    plots=True,
    seed=42               # Ziarno losowości dla reprodukowalności
)
```

```python
# Kontynuacja treningu modelu smoke_detection2
# Załaduj wytrenowany model i kontynuuj trening, zapisując do smoke_detection3

from pathlib import Path
from ultralytics import YOLO

# Ścieżka do checkpoint modelu (używamy last.pt, który zawiera pełny stan treningu)
checkpoint_path = "runs/smoke_detection2/weights/last.pt"

# Sprawdź czy checkpoint istnieje
if not Path(checkpoint_path).exists():
    print(f"⚠️  Checkpoint last.pt nie istnieje: {checkpoint_path}")
    print("   Sprawdzam alternatywne lokalizacje...")
    # Spróbuj best.pt
    checkpoint_path = "runs/smoke_detection2/weights/best.pt"
    if not Path(checkpoint_path).exists():
        raise FileNotFoundError(f"Nie znaleziono checkpoint w runs/smoke_detection2/weights/")
    print(f"   Używam best.pt zamiast last.pt")

print(f"📦 Ładowanie checkpoint: {checkpoint_path}")

# Załaduj model z checkpoint
model = YOLO(checkpoint_path)

print("✅ Model załadowany!")
print(f"🚀 Kontynuuję trening modelu smoke_detection2...")
print(f"   Zapis do: smoke_detection3")

# Ścieżka do pliku data.yaml (powinien istnieć z poprzedniego treningu)
data_yaml = "YOLODataset/data.yaml"
if not Path(data_yaml).exists():
    raise FileNotFoundError(f"Plik data.yaml nie istnieje: {data_yaml}")

# Kontynuuj trening
# epochs - całkowita liczba epok (model był trenowany 30 epok, więc jeśli ustawimy 50, 
# to będzie trenował jeszcze 20 dodatkowych epok)
# Możesz dostosować parametry według potrzeb
results = model.train(
    data=data_yaml,
    epochs=50,          # Łączna liczba epok (30 już było, więc +20 = 50)
    imgsz=640,
    batch=16,
    device=0,
    project="runs",
    name="smoke_detection3",  # Nowa nazwa eksperymentu
    patience=10,
    save=True,
    plots=True,
    verbose=True,
    resume=False  # False - zaczyna nowy eksperyment, ale z wagami z checkpoint
)

print(f"✅ Trening zakończony!")
print(f"📊 Wyniki zapisane w: runs/smoke_detection3")
```
